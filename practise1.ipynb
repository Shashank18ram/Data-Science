```python
num = input("Enter a number: ")

if num == num[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
text = input("Enter a word: ")

text = text.lower()

if text == text[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
arr = [10, 25, 7, 45, 18]

largest = arr[0]

for i in arr:
    if i > largest:
        largest = i

print("Largest element =", largest)
```

```python
arr = [5, 10, 15, 20, 25]

total = 0

for i in arr:
    total += i

print("Sum of array elements =", total)
```
