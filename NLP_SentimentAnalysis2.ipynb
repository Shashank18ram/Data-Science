## VADER — How it works

**VADER (Valence Aware Dictionary and sEntiment Reasoner)** is a rule-based sentiment analysis tool built specifically for social media and review-style text.

At its core, VADER has a **dictionary of words**, each pre-labelled with a sentiment score. For example:
- `"amazing"` → strongly positive
- `"terrible"` → strongly negative
- `"okay"` → mildly positive

But what makes VADER special is that it goes **beyond just word scores**. It understands:

- **Capitalization** → `"AMAZING"` scores higher than `"amazing"`
- **Punctuation** → `"amazing!!!"` scores higher than `"amazing"`
- **Negation** → `"not good"` flips the sentiment of `"good"`
- **Degree modifiers** → `"extremely good"` scores higher than `"kind of good"`

---

**The 4 scores it returns:**

| Score | Meaning |
|-------|---------|
| `pos` | Proportion of text that is positive (0 to 1) |
| `neg` | Proportion of text that is negative (0 to 1) |
| `neu` | Proportion of text that is neutral (0 to 1) |
| `compound` | Overall score normalised between **−1.0 and +1.0** |

The `compound` score is the one we use. Think of it as the final verdict — it combines everything into a single number.

A common threshold:
- `compound > 0` → **positive**
- `compound <= 0` → **negative**

#The steps to vader

Import SentimentIntensityAnalyzer from nltk.sentiment.vader

Run nltk.download('vader_lexicon') — this downloads the word dictionary VADER needs

Create an analyzer object: analyzer = SentimentIntensityAnalyzer()



```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer
lexicon = nltk.download('vader_lexicon')
analyzer = SentimentIntensityAnalyzer()
```

```python
analyzer.polarity_scores('This guy is Pathetic, he just irritates everybody.')
```

```python
analyzer.polarity_scores("this laptop is Amazing!!!")
```

```python
analyzer.polarity_scores('He WAS brilliant, was BEST, was very sharp!!')
```

```python
#Write sentiment analyzer function for each df row
#Give the sentiment, return the compound score
a = lambda txt:analyzer.polarity_scores(txt)['compound']
a('He WAS brilliant, was BEST, was very sharp!!')
```

```python
df_reviews['review_Scores'] = df_reviews['reviews'].apply(a)
df_reviews.head()
```

```python
df_reviews[df_reviews['review_Scores'] <0]
```

#Next Steps
Apply polarity_scores() on every review and extract just the compound score — store it in a new column called "Score".


Based on the score, create another column called "Sentiment" — "positive" if compound > 0, else "negative"

```python
df_reviews['Score'] = df_reviews['reviews'].apply(lambda txt:analyzer.polarity_scores(txt)['compound'])
df_reviews['Sentiment'] = df_reviews['Score'].apply(lambda score: 'Positive' if score>0 else 'Negative')
df_reviews.head(10)
```

```python
#let us clean off all the stuff
df_reviews_clone = df_reviews.copy()
df_reviews_clone.drop(columns=['Score','Sentiment'],inplace=True)
def clean_text(txt):
  import re
  return re.sub(r'[^a-zA-Z0-9\s]','',txt)
df_reviews_clone['reviews'] = df_reviews_clone['reviews'].apply(clean_text)
df_reviews_clone.head()
```

```python
df_reviews_clone['Score'] = df_reviews_clone['reviews'].apply(lambda txt:analyzer.polarity_scores(txt)['compound'])
df_reviews_clone['Sentiment'] = df_reviews_clone['Score'].apply(lambda score: 'Positive' if score>0 else 'Negative')
df_reviews_clone.head()
```

```python
df_reviews = df_reviews_clone
```

```python
df_reviews['Sentiment'].value_counts()
```
