```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
```

```python
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_csv('/content/drive/MyDrive/Datasets/sales_data_with_discounts.csv')
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nDataset Info:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
numerical_cols = df.select_dtypes(include=['int64','float64']).columns

categorical_cols = df.select_dtypes(include=['object']).columns

print("Numerical Columns:")
print(numerical_cols)

print("\nCategorical Columns:")
print(categorical_cols)
```

```python
print(df[numerical_cols].describe())
```

```python
for col in numerical_cols:

    print("="*50)
    print("Column :", col)

    print("Mean :", df[col].mean())

    print("Median :", df[col].median())

    print("Mode :", df[col].mode()[0])

    print("Standard Deviation :", df[col].std())
```

1.Mean
>Represents the average value of the column.

2.Median
>Represents the middle value.
>Less affected by outliers.

3.Mode
>Most frequently occurring value.

4.Standard Deviation
>Shows spread of the data.
>Higher value means more variation.

```python
for col in numerical_cols:

    plt.figure(figsize=(6,4))

    plt.hist(df[col], bins=15, edgecolor='black')

    plt.title("Histogram of " + col)

    plt.xlabel(col)

    plt.ylabel("Frequency")

    plt.show()
```

Histogram Inference
>Symmetric histogram → Normally distributed.
>Right skew → More small values.
>Left skew → More large values.
>Extreme bars → Possible outliers.

```python
for col in numerical_cols:

    plt.figure(figsize=(5,3))

    plt.boxplot(df[col], vert=False)

    plt.title("Boxplot of " + col)

    plt.show()
```

Boxplot Inference
>Dots outside whiskers are outliers.
>Longer whiskers indicate greater spread.
>Small box indicates less variability.

```python
for col in categorical_cols:

    plt.figure(figsize=(7,4))

    df[col].value_counts().plot(kind='bar')

    plt.title("Bar Chart of " + col)

    plt.xlabel(col)

    plt.ylabel("Count")

    plt.show()
```

```python
for col in categorical_cols:

    plt.figure(figsize=(7,4))

    df[col].value_counts().plot(kind='bar')

    plt.title("Bar Chart of " + col)

    plt.xlabel(col)

    plt.ylabel("Count")

    plt.show()
```

Bar Chart Inference
>Highest bar indicates most frequent category.
>Lowest bar indicates least frequent category.
>Helps understand category distribution

```python
print(df[numerical_cols].head())
```

```python
scaler = StandardScaler()

df_standardized = df.copy()

df_standardized[numerical_cols] = scaler.fit_transform(df[numerical_cols])

print(df_standardized.head())
```

```python
df[numerical_cols].describe()
```

```python
df_standardized[numerical_cols].describe()
```

```python
for col in numerical_cols:

    plt.figure(figsize=(6,4))

    plt.hist(df_standardized[col], bins=15, edgecolor='black')

    plt.title("Standardized " + col)

    plt.show()
```

Standardization Explanation

Z-score normalization transforms data using

x = original value
μ = mean
σ = standard deviation

After standardization,

Mean ≈ 0
Standard deviation ≈ 1

This helps machine learning algorithms perform better when features have different scales.
