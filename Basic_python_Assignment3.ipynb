```python
# Exercise 1: Check whether a number is prime or not

num = int(input("Enter a number: "))

if num > 1:
    is_prime = True

    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print(num, "is a Prime Number.")
    else:
        print(num, "is NOT a Prime Number.")
else:
    print(num, "is NOT a Prime Number.")
```

```python
# Exercise 2: Product of two random numbers

import random

num1 = random.randint(1, 10)
num2 = random.randint(1, 10)

answer = int(input(f"What is {num1} × {num2}? "))

if answer == num1 * num2:
    print("Correct! Well done.")
else:
    print("Incorrect!")
    print("Correct Answer:", num1 * num2)
```

```python
# Exercise 3: Print squares of even numbers from 100 to 200
# Choice: Even Numbers

print("Squares of Even Numbers from 100 to 200:\n")

for i in range(100, 201):
    if i % 2 == 0:
        print(f"{i}^2 = {i*i}")
```

```python
# Exercise 4: Word Counter

text = input("Enter a sentence: ")

words = text.split()

for word in words:
    print(word, ":", words.count(word))
```

```python
# Exercise 5: Check Palindrome

word = input("Enter a word: ")

if word == word[::-1]:
    print("True")
else:
    print("False")
```
