```python
import math
from scipy.stats import norm
```

```python
sample_mean = 3050
mu = 1000 + 5 * 600
sigma = 5 * 25
n = 25
alpha = 0.05
```

```python
z = (sample_mean - mu) / (sigma / math.sqrt(n))
```

```python
critical_value = norm.ppf(1 - alpha)
```

```python
print("Theoretical Mean:", mu)
print("Population Standard Deviation:", sigma)
print("Test Statistic (Z):", z)
print("Critical Value:", critical_value)
```

```python
if z > critical_value:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

```python
print("\nConclusion:")
print("There is insufficient evidence to support the claim that weekly operating costs are higher than the model suggests.")
```
